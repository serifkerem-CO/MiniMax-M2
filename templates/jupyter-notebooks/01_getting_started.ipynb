## 1. Kurulum ve Yapılandırma

```python
# Gerekli kütüphaneleri yükle
!pip install openai -q
```

```python
import os
from openai import OpenAI

# API Yapılandırması
API_BASE_URL = "http://localhost:8000/v1"  # veya https://api.minimax.chat/v1
API_KEY = "your-api-key-here"
MODEL_NAME = "MiniMax-M2"

# İstemci oluştur
client = OpenAI(
    api_key=API_KEY,
    base_url=API_BASE_URL,
)

print("✅ İstemci oluşturuldu!")
```

## 2. Temel API Kullanımı

```python
# Basit bir soru sor
response = client.chat.completions.create(
    model=MODEL_NAME,
    messages=[
        {"role": "system", "content": "Sen yardımcı bir asistansın."},
        {"role": "user", "content": "Python nedir? Kısa açıkla."}
    ],
    temperature=0.7,
    max_tokens=500,
)

print(response.choices[0].message.content)
```

```python
# Token kullanımını göster
print(f"Giriş token: {response.usage.prompt_tokens}")
print(f"Çıkış token: {response.usage.completion_tokens}")
print(f"Toplam token: {response.usage.total_tokens}")
```

## 3. Streaming Yanıtlar

Streaming ile yanıtları parça parça alabilirsiniz. Bu, kullanıcı deneyimini iyileştirir.

```python
# Streaming yanıt
stream = client.chat.completions.create(
    model=MODEL_NAME,
    messages=[
        {"role": "user", "content": "1'den 5'e kadar say ve her sayıyı açıkla."}
    ],
    stream=True,
)

print("Streaming yanıt:")
for chunk in stream:
    if chunk.choices[0].delta.content:
        print(chunk.choices[0].delta.content, end="", flush=True)
print()  # Yeni satır
```

## 4. Sohbet Geçmişi

MiniMax-M2, önceki mesajları hatırlayarak bağlamsal yanıtlar verir.

```python
# Sohbet geçmişi ile konuşma
messages = [
    {"role": "system", "content": "Sen yardımcı bir asistansın."},
    {"role": "user", "content": "Benim adım Ali."},
]

# İlk yanıt
response1 = client.chat.completions.create(
    model=MODEL_NAME,
    messages=messages,
)
print("Yanıt 1:", response1.choices[0].message.content)

# Geçmişe ekle
messages.append({"role": "assistant", "content": response1.choices[0].message.content})
messages.append({"role": "user", "content": "Benim adım neydi?"})

# İkinci yanıt
response2 = client.chat.completions.create(
    model=MODEL_NAME,
    messages=messages,
)
print("Yanıt 2:", response2.choices[0].message.content)
```

## 5. Parametre Ayarlama

Farklı parametreler, modelin davranışını etkiler.

```python
# Sıcaklık karşılaştırması
prompt = "Bir şiir yaz."

# Düşük sıcaklık (daha tutarlı)
response_low = client.chat.completions.create(
    model=MODEL_NAME,
    messages=[{"role": "user", "content": prompt}],
    temperature=0.2,
    max_tokens=100,
)

# Yüksek sıcaklık (daha yaratıcı)
response_high = client.chat.completions.create(
    model=MODEL_NAME,
    messages=[{"role": "user", "content": prompt}],
    temperature=0.9,
    max_tokens=100,
)

print("=== Düşük Sıcaklık (0.2) ===")
print(response_low.choices[0].message.content)

print("\n=== Yüksek Sıcaklık (0.9) ===")
print(response_high.choices[0].message.content)
```
